```python
import nltk
from nltk.tokenize import word_tokenize
from nltk import pos_tag

# Download resources
nltk.download('punkt')
nltk.download('punkt_tab')
nltk.download('averaged_perceptron_tagger_eng')

# Input sentence
sentence = "The small boy is playing with a red ball."

# Tokenization
tokens = word_tokenize(sentence)

# POS tagging
tagged_tokens = pos_tag(tokens)

print("Original Sentence:")
print(sentence)

print("\nPOS Tagged Sentence:")
print(tagged_tokens)

# Regex grammar
grammar = r"""
    NP: {<DT>?<JJ>*<NN|NNS>}
    PP: {<IN><NP>}
    VP: {<VB.*><NP|PP>*}
"""

# Create parser
parser = nltk.RegexpParser(grammar)

# Parse
tree = parser.parse(tagged_tokens)

print("\nParsed Tree:")
print(tree)

# Display chunks
print("\nChunks:")

for subtree in tree.subtrees():
    if subtree.label() in ['NP', 'VP', 'PP']:
        phrase = " ".join(word for word, tag in subtree.leaves())
        print(subtree.label(), ":", phrase)

```

```python
import spacy

# Load English language model
nlp = spacy.load("en_core_web_sm")

# Input sentence
sentence = "The small boy is playing with a red ball."

# Process sentence
doc = nlp(sentence)

# Display original sentence
print("Original Sentence:")
print(sentence)

# Tokenization
print("\nTokens:")
for token in doc:
    print(token.text)

# POS Tagging
print("\nPOS Tagging:")
for token in doc:
    print(token.text, "->", token.pos_, "->", token.tag_)

# Noun Chunking
print("\nNoun Chunks:")
for chunk in doc.noun_chunks:
    print(chunk.text)

# Dependency Parsing
print("\nDependency Parsing:")
for token in doc:
    print(
        token.text,
        "-> Head:", token.head.text,
        "-> Dependency:", token.dep_
    )

# Detailed information
print("\nDetailed Parsing:")
for token in doc:
    print(
        f"Word: {token.text:10} "
        f"POS: {token.pos_:8} "
        f"Tag: {token.tag_:5} "
        f"Dependency: {token.dep_:10} "
        f"Head: {token.head.text}"
    )
```
