```python
!wget -q https://nlp.stanford.edu/data/glove.6B.zip
!unzip -q glove.6B.zip

import numpy as np

embeddings = {}

with open("glove.6B.100d.txt", encoding="utf8") as f:
    for line in f:
        values = line.split()
        word = values[0]
        vector = np.asarray(values[1:], dtype="float32")
        embeddings[word] = vector

word = "machine"

print("Vector for:", word)
print(embeddings[word])
print("\nVector dimension:", len(embeddings[word]))
```

```python
from sklearn.metrics.pairwise import cosine_similarity

word = "machine"
vector = embeddings[word].reshape(1, -1)

similarities = {}

for w, v in embeddings.items():
    similarities[w] = cosine_similarity(
        vector, v.reshape(1, -1)
    )[0][0]

similar_words = sorted(
    similarities.items(),
    key=lambda x: x[1],
    reverse=True
)[1:6]

print(similar_words)
```
