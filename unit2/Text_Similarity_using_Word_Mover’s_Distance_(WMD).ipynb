```python

# Text Similarity using Word Mover's Distance (WMD)

!pip install -q gensim nltk POT

import nltk
nltk.download('punkt', quiet=True)
nltk.download('punkt_tab', quiet=True)
nltk.download('stopwords', quiet=True)

import numpy as np
from gensim.models import Word2Vec
from gensim.parsing.preprocessing import STOPWORDS
from nltk.tokenize import word_tokenize
from nltk.corpus import stopwords
import re

# Sample documents
documents = [
    "Machine learning is a branch of artificial intelligence",
    "Artificial intelligence uses machine learning algorithms",
    "The weather is pleasant and sunny today",
    "Deep learning is used in modern artificial intelligence"
]

# Text preprocessing
stop_words = set(stopwords.words('english'))

def preprocess(text):
    text = re.sub(r'[^a-zA-Z\s]', '', text.lower())
    tokens = word_tokenize(text)
    return [word for word in tokens if word not in stop_words]

tokenized_docs = [preprocess(doc) for doc in documents]

# Train Word2Vec model
model = Word2Vec(
    sentences=tokenized_docs,
    vector_size=100,
    window=5,
    min_count=1,
    workers=4,
    epochs=100
)

# Calculate Word Mover's Distance
def calculate_wmd(doc1, doc2):
    return model.wv.wmdistance(doc1, doc2)

# Convert WMD distance into similarity score
def similarity_score(distance):
    if np.isinf(distance):
        return 0
    return 1 / (1 + distance)

# Compare all documents
print("TEXT SIMILARITY USING WORD MOVER'S DISTANCE\n")

for i in range(len(tokenized_docs)):
    for j in range(i + 1, len(tokenized_docs)):
        distance = calculate_wmd(tokenized_docs[i], tokenized_docs[j])
        similarity = similarity_score(distance)

        print(f"Document {i+1} vs Document {j+1}")
        print(f"WMD Distance : {distance:.4f}")
        print(f"Similarity   : {similarity:.4f}")
        print("-" * 45)

# Compare two custom sentences
text1 = "Natural language processing helps computers understand human language"
text2 = "NLP enables machines to interpret human communication"

tokens1 = preprocess(text1)
tokens2 = preprocess(text2)

distance = calculate_wmd(tokens1, tokens2)

print("\nCUSTOM TEXT COMPARISON")
print("Text 1:", text1)
print("Text 2:", text2)
print(f"WMD Distance: {distance:.4f}")
print(f"Similarity Score: {similarity_score(distance):.4f}")

```
