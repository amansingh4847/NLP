punkt is a pretrained model.

nltk - natural lang toolkit.

```python
import nltk
nltk.download('punkt')
from nltk.tokenize import sent_tokenize, word_tokenize
```

```python
text = "Natural language Processing is Interesting. It is used in AI."
```

```python
nltk.download('punkt_tab')
sentences = sent_tokenize(text)
for sentence in sentences:
  print(sentence)
```

```python
nltk.download('punkt_tab')
words = word_tokenize(text)
count = 0
for word in words:
  count += 1
  print(word)
print("Total No of tokens = ",count)
```

```python
import spacy #Model like punkt
nlp = spacy.load('en_core_web_sm')
```

```python
doc = nlp(text)
```

```python
for sent in doc.sents:
  print(sent.text)
```

```python
for token in doc:
  print(token.text)
```
